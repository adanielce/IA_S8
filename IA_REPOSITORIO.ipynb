## Regresión Lineal Simple


```python
def regresion_lineal_simple(X, Y):
    n = len(X)
    
    if n != len(Y):
        print("Error: La cantidad de datos en X no es igual a la cantidad en Y.")
        return

    # Calcular sumatorias
    sum_x = sum(X)
    sum_y = sum(Y)
    sum_xy = sum(x * y for x, y in zip(X, Y))
    sum_x2 = sum(x ** 2 for x in X)
    sum_y2 = sum(y ** 2 for y in Y)
    
    # Imprimir tabla paso a paso
    print("--- TABLA DE CÁLCULOS ---")
    print(f"{'X (Hora)':<10} | {'Y (Temp)':<10} | {'X*Y':<10} | {'X^2':<10} | {'Y^2':<10}")
    print("-" * 60)
    for i in range(n):
        print(f"{X[i]:<10.2f} | {Y[i]:<10.2f} | {X[i]*Y[i]:<10.2f} | {X[i]**2:<10.2f} | {Y[i]**2:<10.2f}")
    print("-" * 60)
    print(f"ΣX={sum_x:.2f} | ΣY={sum_y:.2f} | ΣXY={sum_xy:.2f} | ΣX²={sum_x2:.2f} | ΣY²={sum_y2:.2f}\n")
    
    # Calcular m (pendiente) y b (intersección)
    numerador_m = (n * sum_xy) - (sum_x * sum_y)
    denominador_m = (n * sum_x2) - (sum_x ** 2)
    m = numerador_m / denominador_m
    
    b = (sum_y - (m * sum_x)) / n
    
    # Resultados
    print("\n--- RESULTADOS DEL MODELO ---")
    print(f"Ecuación de la recta: Y = ({m:.4f} * X) + {b:.4f}")
    
    # Interacción con el usuario para estimar la temperatura
    print("\n" + "="*50)
    print("ESTIMADOR DE TEMPERATURA")
    print("="*50)
    
    while True:
        try:
            entrada = input("\n¿De qué hora quieres saber la temperatura? (escribe 'salir' para terminar): ")
            if entrada.strip().lower() == 'salir':
                print("¡Programa terminado!")
                break
            
            hora_estimar = float(entrada)
            # Aplicamos la fórmula y = mx + b
            temperatura_estimada = (m * hora_estimar) + b
            
            print(f"👉 A la hora {hora_estimar}, la temperatura estimada es de: {temperatura_estimada:.2f} °C")
            
        except ValueError:
            print("❌ Entrada inválida. Por favor, ingresa un número de hora válido (ej. 9 o 14.5).")



X_datos = [12,14,16,18,20,22,24,2,4,6] 
Y_datos = [26.88,25.42,24.93,24.44,23.95,22.48,21.99,21.51,19.55,19.06] 

# Ejecutar función
regresion_lineal_simple(X_datos, Y_datos)
```

## Sistema Experto - Factores de Certeza


```python
class SistemaExpertoVista:
    def __init__(self):
        self.hechos = {}
        self.reglas = []
        self.conclusiones = {}
        self.traza = []

    def definir_reglas(self):
        self.reglas = [
            {'id': 1, 'premisas': [('vision_borrosa_lejos', True), ('fatiga_visual', True)], 'conclusion': 'Posible_Miopia', 'certeza': 0.85},
            {'id': 2, 'premisas': [('vision_borrosa_cerca', True), ('fatiga_visual', True)], 'conclusion': 'Posible_Hipermetropia', 'certeza': 0.85},
            {'id': 3, 'premisas': [('destellos_luz', True), ('manchas_flotantes', True)], 'conclusion': 'Alerta_Retina', 'certeza': 0.90},
            {'id': 4, 'premisas': [('sensibilidad_luz', True), ('picazon_ojos', True)], 'conclusion': 'Posible_Conjuntivitis', 'certeza': 0.80},
            {'id': 5, 'premisas': [('Posible_Miopia', True), ('dolor_cabeza', True)], 'conclusion': 'Miopia_Sintomatica_Requiere_Lentes', 'certeza': 0.95},
            {'id': 6, 'premisas': [('Alerta_Retina', True), ('vision_borrosa_lejos', True)], 'conclusion': 'Urgencia_Desprendimiento_Retina', 'certeza': 0.95}
        ]

    def obtener_sintomas(self):
    #interfaz entrada
        print("\n=== Evaluación de vision ===")
        
        # hechos de los sintomas 
        self.hechos = {
            'vision_borrosa_lejos': True,
            'vision_borrosa_cerca': False,
            'dolor_cabeza': True,
            'fatiga_visual': True,
            'destellos_luz': False,
            'manchas_flotantes': False,
            'sensibilidad_luz': False,
            'picazon_ojos': False
        }
        for síntoma, valor in self.hechos.items():
            print(f"[{'X' if valor else ' '}] {síntoma.replace('_', ' ').title()}")

   
    def obtener_valor_hecho(self, nombre_hecho):
        if nombre_hecho in self.hechos:
            return self.hechos[nombre_hecho], 1.0
        if nombre_hecho in self.conclusiones:
            return True, self.conclusiones[nombre_hecho]
        return False, 0.0

    def motor_inferencia(self):
#evalua reglas y certezas
        cambio = True
        while cambio:
            cambio = False
            for regla in self.reglas:
                if regla['conclusion'] in self.conclusiones:
                    continue
                
                premisas_cumplidas = True
                certeza_minima = 1.0
                detalles_premisas = []
                
                for nombre, valor_esperado in regla['premisas']:
                    valor_real, certeza_hecho = self.obtener_valor_hecho(nombre)
                    if valor_esperado == True and valor_real == True:
                        detalles_premisas.append(f"{nombre}={valor_real}")
                        certeza_minima = min(certeza_minima, certeza_hecho)
                    elif valor_esperado == False and valor_real == False:
                        detalles_premisas.append(f"{nombre}={valor_real}")
                    else:
                        premisas_cumplidas = False
                        break
                        
                if premisas_cumplidas:
                    #regla*certeza
                    certeza_final = regla['certeza'] * certeza_minima
                    self.conclusiones[regla['conclusion']] = certeza_final
                    

                    self.traza.append({
                        'regla': regla['id'], 
                        'conclusion': regla['conclusion'], 
                        'certeza_base': regla['certeza'],
                        'certeza_final': certeza_final, 
                        'premisas': detalles_premisas
                    })
                    cambio = True

    def mostrar_resultados(self):
        print("\n" + "="*50)
        print("RESULTADOS DEL DIAGNÓSTICO ")
        print("="*50)
        print("\n--- CONCLUSIONES ---")
        
        if not self.conclusiones:
            print("No se detectaron problemas .")
        else:
            ordenadas = sorted(self.conclusiones.items(), key=lambda x: x[1], reverse=True)
            for concl, certeza in ordenadas:
                porcentaje = round(certeza * 100, 2)
                print(f"-> {concl.replace('_', ' ')}: {porcentaje}% de probabilidad")

        print("\n--- TRAZABILIDAD ---")
        for paso in self.traza:
            print(f"\nRegla #{paso['regla']} activada:")
            print(f"  Diagnóstico: {paso['conclusion']}")
            print(f"  Basado en: {', '.join(paso['premisas'])}")
            print(f"  Cálculo: Certeza Regla ({paso['certeza_base']}) x Certeza Premisa = {paso['certeza_final']:.3f}")

if __name__ == "__main__":
    se_vista = SistemaExpertoVista()
    se_vista.definir_reglas()
    se_vista.obtener_sintomas()
    se_vista.motor_inferencia()
    se_vista.mostrar_resultados()
```

## Práctica 1


```python
#PRACTICA 1 ADCE
def pre():
    preguntas = {
        "fiebre": "¿Tienes fiebre?",
        "dolor_cuerpo": "¿Tienes dolor de cuerpo?",
        "tos": "¿Tienes tos?",
        "dolor_garganta": "¿Tienes dolor de garganta?",
        "fatiga": "¿Sientes fatiga extrema?"
    }

    hechos = set()
    print("--- Sistema de Diagnóstico ---")
    
    for clave, pregunta in preguntas.items():
        respuesta = input(f"{pregunta} (si/no): ").lower()
        if respuesta == 'si':
            hechos.add(clave)

    reglas = [
        ({"fiebre", "dolor_cuerpo"}, "Gripe"),
        ({"tos", "dolor_garganta"}, "Resfriado"),
        ({"fiebre", "tos", "fatiga"}, "Infección Viral"),
        ({"dolor_garganta", "fiebre"}, "Infección de Garganta")
    ]

    print(f"\nHechos ingresados (Axiomas): {list(hechos)}")
    
    hay_conclusion = False
    for síntomas, resultado in reglas:
        if síntomas.issubset(hechos):
            print(f"\nResultado (Teorema): {resultado}")
            print(f"Demostración: Se concluye '{resultado}' porque se cumplen los hechos {list(síntomas)}")
            hay_conclusion = True

    if not hay_conclusion:
        print("\nNo se encontró un diagnóstico con esos síntomas.")

if __name__ == "__main__":
    pre()
```

```python
def inferir(hechos, reglas):
    conclusiones = []
    pasos = []
    reglas_activadas = 0
    
    for condiciones, conclusion in reglas:
        if all(condicion in hechos for condicion in condiciones):
            conclusiones.append(conclusion)
            reglas_activadas += 1
            
            condiciones_str = " y ".join(condiciones)
            pasos.append(f"Porque tienes {condiciones_str}, puede ser {conclusion}.")
            
    return conclusiones, pasos, reglas_activadas

hechos = {"fiebre", "tos", "dolor_cuerpo"}

reglas = [
    (["fiebre", "dolor_cuerpo"], "gripe"),
    (["fiebre", "tos"], "gripe")
]

conclusiones, pasos, total_reglas = inferir(hechos, reglas)

print("Hechos detectados:", hechos)
print("Conclusiones:", conclusiones)

print("\nDemostración:")
for paso in pasos:
    print("-", paso)

print(f"\nSe activaron {total_reglas} reglas")
```

# 🥗 Proyecto - Sistema Experto con Manejo de Incertidumbre (Nutrición)

## Sistema Experto Nutrición


```python
class SistemaExpertoNutricion:
    def __init__(self):
        self.hechos = {}
        self.reglas = []
        self.conclusiones = {}
        self.traza = []
        self.edad = 0
        self.peso = 0.0
        self.estatura = 0.0
        self.imc = 0.0

    def definir_reglas(self):
        self.reglas = [
            {'id': 1, 'premisas': [('busca_ganar_musculo', True), ('entrena_pesas', True)], 'conclusion': 'Fase_De_Volumen', 'certeza': 0.90},
            {'id': 2, 'premisas': [('busca_perder_peso', True), ('hace_cardio', True)], 'conclusion': 'Deficit_Calorico', 'certeza': 0.85},
            {'id': 3, 'premisas': [('poco_tiempo_cocinar', True), ('presupuesto_limitado', True)], 'conclusion': 'Perfil_Practico_Economico', 'certeza': 0.80},
            {'id': 4, 'premisas': [('desayuna_pesado', False), ('hace_cardio', True)], 'conclusion': 'Entrenamiento_Ayunas_Apto', 'certeza': 0.75},
            {'id': 5, 'premisas': [('Fase_De_Volumen', True), ('intolerancia_lactosa', True)], 'conclusion': 'Dieta_Hipercalorica_Sin_Lacteos', 'certeza': 0.95},
            {'id': 6, 'premisas': [('Deficit_Calorico', True), ('Perfil_Practico_Economico', True)], 'conclusion': 'Dieta_Economica_De_Preparacion_Rapida', 'certeza': 0.90}
        ]

    def obtener_datos_usuario(self):
        print("\n" + "="*55)
        print("CUESTIONARIO BIOMÉTRICO Y DE NUTRICIÓN")
        print("="*55)
        
        # Validación estricta para la edad
        while True:
            try:
                self.edad = int(input("¿Cuál es tu edad en años? (ej. 22): "))
                if self.edad <= 0:
                    raise ValueError
                break
            except ValueError:
                print(" Entrada inválida. Por favor, ingresa un número entero válido mayor a 0.\n")

        # Validación estricta para el peso
        while True:
            try:
                self.peso = float(input("¿Cuál es tu peso en kg? (ej. 75.5): "))
                if self.peso <= 0:
                    raise ValueError
                break
            except ValueError:
                print(" Entrada inválida. Por favor, ingresa un número válido mayor a 0 (usa punto para decimales).\n")

        # Validación estricta para la estatura
        while True:
            try:
                self.estatura = float(input("¿Cuál es tu estatura en metros? (ej. 1.75): "))
                if self.estatura <= 0:
                    raise ValueError
                break
            except ValueError:
                print(" Entrada inválida. Por favor, ingresa un número válido mayor a 0 (usa punto para decimales).\n")

        self.imc = self.peso / (self.estatura ** 2)

        print("\nPor favor, responde con 's' (sí) o 'n' (no) a las siguientes preguntas:\n")
        preguntas = {
            'busca_perder_peso': '¿Buscas perder peso o quemar grasa?',
            'busca_ganar_musculo': '¿Buscas ganar masa muscular?',
            'entrena_pesas': '¿Entrenas con pesas o haces calistenia?',
            'hace_cardio': '¿Realizas ejercicio cardiovascular regularmente?',
            'intolerancia_lactosa': '¿Tienes intolerancia a la lactosa?',
            'presupuesto_limitado': '¿Tienes un presupuesto limitado para comidas?',
            'poco_tiempo_cocinar': '¿Dispones de poco tiempo para cocinar?',
            'desayuna_pesado': '¿Acostumbras desayunar porciones abundantes?'
        }
        
        for clave, pregunta in preguntas.items():
            while True:
                respuesta = input(f"{pregunta} (s/n): ").strip().lower()
                if respuesta in ['s', 'si', 'n', 'no']:
                    self.hechos[clave] = True if respuesta in ['s', 'si'] else False
                    break
                else:
                    print(" Por favor, responde únicamente con 's' o 'n'.")

    def obtener_valor_hecho(self, nombre_hecho):
        if nombre_hecho in self.hechos:
            return self.hechos[nombre_hecho], 1.0
        if nombre_hecho in self.conclusiones:
            return True, self.conclusiones[nombre_hecho]
        return False, 0.0

    def motor_inferencia(self):
        cambio = True
        while cambio:
            cambio = False
            for regla in self.reglas:
                if regla['conclusion'] in self.conclusiones:
                    continue
                
                premisas_cumplidas = True
                certeza_minima = 1.0
                
                for nombre, valor_esperado in regla['premisas']:
                    valor_real, certeza_hecho = self.obtener_valor_hecho(nombre)
                    if valor_esperado == valor_real:
                        if valor_real == True:
                            certeza_minima = min(certeza_minima, certeza_hecho)
                    else:
                        premisas_cumplidas = False
                        break
                        
                if premisas_cumplidas:
                    certeza_final = regla['certeza'] * certeza_minima
                    self.conclusiones[regla['conclusion']] = certeza_final
                    self.traza.append({'regla': regla['id'], 'conclusion': regla['conclusion'], 'certeza_final': certeza_final})
                    cambio = True

    def mostrar_resultados(self):
        print("\n" + "="*55)
        print("REPORTE Y RECOMENDACIONES PERSONALIZADAS")
        print("="*55)
        
        categoria_imc = ""
        if self.imc < 18.5:
            categoria_imc = "Bajo peso"
        elif 18.5 <= self.imc < 24.9:
            categoria_imc = "Peso normal"
        elif 25 <= self.imc < 29.9:
            categoria_imc = "Sobrepeso"
        else:
            categoria_imc = "Obesidad"

        print(f"\n--- PERFIL BIOMÉTRICO ---")
        print(f"Edad: {self.edad} años")
        print(f"Peso: {self.peso} kg | Estatura: {self.estatura} m")
        print(f"Tu IMC es: {self.imc:.2f} ({categoria_imc})")

        print("\n--- PLAN SUGERIDO DEL SISTEMA EXPERTO ---")
        if not self.conclusiones:
            print("Faltan datos para una conclusión específica.")
        else:
            ordenadas = sorted(self.conclusiones.items(), key=lambda x: x[1], reverse=True)
            for concl, certeza in ordenadas:
                print(f"-> {concl.replace('_', ' ')}: {round(certeza * 100, 2)}% de viabilidad")

        print("\n--- RECOMENDACIONES FINALES ---")
        if "Deficit_Calorico" in self.conclusiones:
            if self.imc >= 25:
                print("* Excelente decisión buscar un déficit calórico. Tu IMC indica que reducir tu porcentaje de grasa beneficiará tu salud general.")
            else:
                print("* CUIDADO: Buscas perder peso, pero tu IMC ya es bajo o normal. Te recomendamos enfocarte en recomposición corporal (comer mantenimiento y entrenar fuerza) en lugar de bajar drásticamente tus calorías.")
                
        if "Fase_De_Volumen" in self.conclusiones:
            if self.imc < 24.9:
                print("* Tienes un IMC ideal para empezar una fase de volumen (superávit calórico). Aumenta tus carbohidratos gradualmente para tener energía en tus entrenamientos.")
            else:
                print("* NOTA: Tu objetivo es ganar masa muscular, pero tu IMC indica sobrepeso. Considera primero hacer una pequeña fase de definición (déficit ligero) antes de subir tus calorías, para no ganar exceso de grasa.")
                
        if "Dieta_Economica_De_Preparacion_Rapida" in self.conclusiones:
            print("* Para tu perfil económico y rápido, te sugerimos basar tus comidas en: atún, huevos, pollo deshebrado, avena, arroz y vegetales congelados. Puedes cocinar todo el domingo (Meal Prep).")

if __name__ == "__main__":
    se_nutricion = SistemaExpertoNutricion()
    se_nutricion.definir_reglas()
    se_nutricion.obtener_datos_usuario()
    se_nutricion.motor_inferencia()
    se_nutricion.mostrar_resultados()
```
